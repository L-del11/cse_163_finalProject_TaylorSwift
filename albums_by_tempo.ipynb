```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv("taylor_album_songs.csv")

tempo_by_album = df.groupby("album_name")["tempo"].mean().reset_index()

sns.relplot(data = tempo_by_album, x = "album_name", y= "tempo", kind = "line")
plt.xticks(rotation = 65)
plt.xlabel("Album Name")
plt.ylabel("Average Tempo")
plt.title("Average Tempo per Taylor Swift Album")

plt.savefig('avg_tempo_by_album.png', bbox_inches='tight')
```
